```python
import pandas as pd
df = pd.read_csv('nvidia_stock_data_1999_2026.csv')
df.head() # This will show the first 5 rows
```

```python
import plotly.express as px

# 1. Revenue vs Market Cap (Fundamental Check)
# 2. RSI Analysis (Technical Check)
fig_rsi = px.line(df, x='date', y='rsi_14', title="RSI-14 (Overbought > 70, Oversold < 30)")
fig_rsi.add_hline(y=70, line_dash="dash", line_color="red")
fig_rsi.add_hline(y=30, line_dash="dash", line_color="green")
# 3. Key Event Impact (Event Check)
events = df[df['key_event'].notna()]
print("Major Impact Events:")
print(events[['date', 'key_event', 'close', 'volume']])
```
